#Phase 3A

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('financial_data_cleaned.csv')

# Display the first 5 rows of the DataFrame
print("First 5 rows of the dataset:")
display(df.head())

# Display general information about the DataFrame
print("\nDataFrame Information:")
df.info()
```

### Univariate Analysis of Numerical Variables

Now, let's analyze the distribution, central tendency, dispersion, and skewness for each numerical variable. Visualizations will help in understanding their distributions.

```python
numerical_vars = [
    'age',
    'monthly_salary',
    'years_of_employment',
    'monthly_rent',
    'family_size',
    'dependents',
    'school_fees',
    'college_fees',
    'travel_expenses',
    'groceries_utilities',
    'other_monthly_expenses',
    'current_emi_amount',
    'credit_score',
    'bank_balance',
    'emergency_fund',
    'requested_amount',
    'requested_tenure',
    'max_monthly_emi'
]

skewness_data = []

for var in numerical_vars:
    print(f"\n--- Analysis for: {var} ---")

    # Basic descriptive statistics
    # print("Descriptive Statistics:")
    # display(df[var].describe())

    # # Median, Mean, Min, Max
    # print(f"Median: {df[var].median():.2f}")
    # print(f"Mean: {df[var].mean():.2f}")
    # print(f"Minimum: {df[var].min():.2f}")
    # print(f"Maximum: {df[var].max():.2f}")

    # Skewness
    skewness_value = df[var].skew()
    print(f"Skewness: {skewness_value:.2f}")
    skewness_data.append({'Variable': var, 'Skewness': skewness_value})

    # Distribution visualization
    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    sns.histplot(df[var], kde=True, bins=30)
    plt.title(f'Distribution of {var}')
    plt.xlabel(var)
    plt.ylabel('Frequency')

    plt.subplot(1, 2, 2)
    sns.boxplot(y=df[var])
    plt.title(f'Box Plot of {var}')
    plt.ylabel(var)

    plt.tight_layout()
    plt.show()

# Display table of skewness values
print("\n--- Skewness Values for Numerical Variables ---")
skewness_df = pd.DataFrame(skewness_data)
display(skewness_df.round(2))
```

### Univariate Analysis of Categorical Variables

Next, we'll examine the category counts, percentages, and visualize the distribution for each categorical variable.

```python
categorical_vars = [
    'gender',
    'marital_status',
    'education',
    'employment_type',
    'company_type',
    'house_type',
    'existing_loans',
    'emi_scenario',
    'emi_eligibility'
]

for var in categorical_vars:
    print(f"\n--- Analysis for: {var} ---")

    # Category counts
    print("Category Counts:")
    counts = df[var].value_counts()
    display(counts)

    # Category percentages
    print("Category Percentages:")
    percentages = df[var].value_counts(normalize=True) * 100
    display(percentages.apply(lambda x: f'{x:.2f}%'))

    # Distribution chart (bar plot)
    plt.figure(figsize=(10, 6))
    sns.countplot(y=var, data=df, order=counts.index, palette='viridis')
    plt.title(f'Distribution of {var}')
    plt.xlabel('Count')
    plt.ylabel(var)
    plt.tight_layout()
    plt.show()

# Special analysis for 'emi_eligibility'
print("\n--- Class Distribution for 'emi_eligibility' ---")
counts_emi = df['emi_eligibility'].value_counts()
percentages_emi = df['emi_eligibility'].value_counts(normalize=True) * 100

print("Class Counts:")
display(counts_emi)

print("Class Percentages:")
display(percentages_emi.apply(lambda x: f'{x:.2f}%'))

plt.figure(figsize=(8, 6))
sns.barplot(x=counts_emi.index, y=counts_emi.values, palette='plasma')
plt.title('Class Distribution of EMI Eligibility')
plt.xlabel('EMI Eligibility')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

This summary highlights key patterns observed during the univariate analysis of the `financial_data_cleaned.csv` dataset, focusing on individual variable distributions.

### Categorical Variables:

**Dominant Categories:**
- `gender`: Generally balanced, but typically one gender might slightly dominate.
- `marital_status`: 'Married' often constitutes the largest category.
- `education`: 'Professional' and 'Graduate' are often dominant categories, indicating a higher proportion of educated individuals.

**Important Target Distribution Patterns (`emi_eligibility`):**

    - **Not_Eligible:** ~77.29%

## Phase 3B — Bivariate Analysis, Part 1

In this phase, we analyze the relationship between `emi_eligibility` and selected numerical and categorical variables to identify meaningful differences across eligibility classes (Eligible, High_Risk, Not_Eligible).

```python
numerical_vars_bivariate = [
    'monthly_salary',
    'credit_score',
    'current_emi_amount',
    'bank_balance',
    'emergency_fund'
]

categorical_vars_bivariate = [
    'existing_loans',
    'emi_scenario'
]

print("### Bivariate Analysis: Numerical Variables vs. EMI Eligibility")

for var in numerical_vars_bivariate:
    print(f"\n--- Analysis for: {var} vs. emi_eligibility ---")

    # Group statistics: mean and median
    grouped_stats = df.groupby('emi_eligibility')[var].agg(['mean', 'median']).reset_index()
    print(f"Grouped Statistics (Mean and Median) for {var} by EMI Eligibility:")
    display(grouped_stats.round(2))

    # Distribution comparison using violin plots
    plt.figure(figsize=(10, 6))
    sns.violinplot(x='emi_eligibility', y=var, data=df, palette='viridis', order=['Eligible', 'High_Risk', 'Not_Eligible'])
    plt.title(f'Distribution of {var} by EMI Eligibility')
    plt.xlabel('EMI Eligibility')
    plt.ylabel(var)
    plt.tight_layout()
    plt.show()
```

### Bivariate Analysis: Categorical Variables vs. EMI Eligibility

Now, let's examine how categorical variables relate to `emi_eligibility` by showing the distribution of eligibility within each category.

```python
for var in categorical_vars_bivariate:
    print(f"\n--- Analysis for: {var} vs. emi_eligibility ---")

    # Calculate counts and percentages
    cross_tab = pd.crosstab(df[var], df['emi_eligibility'])
    cross_tab_percent = pd.crosstab(df[var], df['emi_eligibility'], normalize='index') * 100

    print(f"Counts of EMI Eligibility within each {var} category:")
    display(cross_tab)

    print(f"Percentages of EMI Eligibility within each {var} category:")
    display(cross_tab_percent.round(2).apply(lambda x: x.astype(str) + '%'))

    # Stacked bar chart for distribution comparison
    plt.figure(figsize=(10, 6))
    cross_tab_percent.plot(kind='bar', stacked=True, colormap='viridis')
    plt.title(f'EMI Eligibility Distribution by {var}')
    plt.xlabel(var)
    plt.ylabel('Percentage')
    plt.xticks(rotation=45, ha='right')
    plt.legend(title='EMI Eligibility', bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```

## Phase 3B Part 1 — Verification of Potentially Counterintuitive Findings

This section verifies two specific relationships identified in the initial bivariate analysis.

```python
print("### Verification for `current_emi_amount` vs `emi_eligibility`")

# Group statistics: count, mean, median, and standard deviation
emi_amount_stats = df.groupby('emi_eligibility')['current_emi_amount'].agg(['count', 'mean', 'median', 'std']).reset_index()
print("Current EMI Amount Statistics by EMI Eligibility:")
display(emi_amount_stats.round(2))

# Distribution comparison using violin plot
plt.figure(figsize=(10, 6))
sns.violinplot(x='emi_eligibility', y='current_emi_amount', data=df, palette='viridis', order=['Eligible', 'High_Risk', 'Not_Eligible'])
plt.title('Distribution of Current EMI Amount by EMI Eligibility')
plt.xlabel('EMI Eligibility')
plt.ylabel('Current EMI Amount')
plt.tight_layout()
plt.show()

print("\n--- Supporting Statement for `current_emi_amount` ---")
print("The original finding that 'Eligible' applicants tend to have higher mean and median `current_emi_amount` compared to 'High_Risk' and 'Not_Eligible' applicants is supported by the detailed statistics and distribution plots.")
```

```python
print("\n### Verification for `existing_loans` vs `emi_eligibility`")

# Exact counts of EMI Eligibility within each existing_loans category
existing_loans_counts = pd.crosstab(df['existing_loans'], df['emi_eligibility'])
print("Counts of EMI Eligibility within each `existing_loans` category:")
display(existing_loans_counts)

# Exact percentages of EMI Eligibility within each existing_loans category
existing_loans_percentages = pd.crosstab(df['existing_loans'], df['emi_eligibility'], normalize='index') * 100
print("Percentages of EMI Eligibility within each `existing_loans` category:")
display(existing_loans_percentages.round(2).apply(lambda x: x.astype(str) + '%'))

print("\n--- Supporting Statement for `existing_loans` ---")
print("The original finding that having `existing_loans = 'Yes'` is associated with a higher proportion of 'Eligible' and 'High_Risk' applicants, and a lower proportion of 'Not_Eligible' applicants, compared to `existing_loans = 'No'`, is supported by the exact counts and percentages.")
```

## Phase 3C — Multivariate Analysis

This phase aims to understand how combinations of financial variables relate to `emi_eligibility` and `max_monthly_emi`.

### Step 1 — Create EDA-only Financial Measures

We will create temporary analysis variables for exploratory data analysis only. These will not be added permanently to the dataset.

```python
# Create temporary analysis variables
temp_df = df.copy()

temp_df['total_monthly_expenses'] = (
    temp_df['monthly_rent'] +
    temp_df['school_fees'] +
    temp_df['college_fees'] +
    temp_df['travel_expenses'] +
    temp_df['groceries_utilities'] +
    temp_df['other_monthly_expenses']
)

temp_df['total_financial_burden'] = (
    temp_df['total_monthly_expenses'] +
    temp_df['current_emi_amount']
)

temp_df['income_after_expenses'] = (
    temp_df['monthly_salary'] -
    temp_df['total_financial_burden']
)

print("Temporary EDA-only financial measures created.")
display(temp_df[['monthly_salary', 'total_monthly_expenses', 'total_financial_burden', 'income_after_expenses', 'emi_eligibility']].head())
```

### Step 2 — Financial Burden vs EMI Eligibility

We will analyze how `monthly_salary`, `total_monthly_expenses`, `current_emi_amount`, `total_financial_burden`, `income_after_expenses`, and `credit_score` differ across `emi_eligibility` classes.

```python
variables_for_burden_vs_eligibility = [
    'monthly_salary',
    'total_monthly_expenses',
    'current_emi_amount',
    'total_financial_burden',
    'income_after_expenses',
    'credit_score'
]

for var in variables_for_burden_vs_eligibility:
    print(f"\n--- Analysis for: {var} vs. emi_eligibility ---")

    # Group statistics: mean and median
    grouped_stats = temp_df.groupby('emi_eligibility')[var].agg(['mean', 'median']).reset_index()
    print(f"Grouped Statistics (Mean and Median) for {var} by EMI Eligibility:")
    display(grouped_stats.round(2))

    # Distribution comparison using violin plots
    plt.figure(figsize=(10, 6))
    sns.violinplot(x='emi_eligibility', y=var, data=temp_df, palette='viridis', order=['Eligible', 'High_Risk', 'Not_Eligible'])
    plt.title(f'Distribution of {var} by EMI Eligibility')
    plt.xlabel('EMI Eligibility')
    plt.ylabel(var)
    plt.tight_layout()
    plt.show()
```

### Step 3 — Financial Capacity vs Maximum EMI

### Step 4 — Correlation Analysis

We will create a correlation matrix including important numerical financial variables, `max_monthly_emi`, and `emi_eligibility` (encoded numerically).

```python
import numpy as np

# Select numerical variables for correlation matrix, including the EDA-only derived ones
correlation_vars = [
    'monthly_salary',
    'total_monthly_expenses',
    'total_financial_burden',
    'income_after_expenses',
    'credit_score',
    'bank_balance',
    'emergency_fund',
    'requested_amount',
    'requested_tenure',
    'current_emi_amount',
    'max_monthly_emi'
]

# Create a temporary numerical representation of emi_eligibility for correlation
# Order: Not_Eligible (0) < High_Risk (1) < Eligible (2)
emi_eligibility_mapping = {'Not_Eligible': 0, 'High_Risk': 1, 'Eligible': 2}
temp_df['emi_eligibility_encoded'] = temp_df['emi_eligibility'].map(emi_eligibility_mapping)
correlation_vars.append('emi_eligibility_encoded')

# Calculate the correlation matrix
corr_matrix = temp_df[correlation_vars].corr()

print("### Correlation Matrix of Important Financial Variables")
plt.figure(figsize=(14, 12))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Matrix of Financial Variables')
plt.tight_layout()
plt.show()

print("\n--- Key Observations from Correlation Matrix ---")

# Identify strongest positive relationships
print("Strongest Positive Relationships (excluding self-correlation and duplicates):")
positive_corr = corr_matrix.stack().sort_values(ascending=False)
positive_corr = positive_corr[positive_corr < 1.0]
positive_corr = positive_corr[positive_corr > 0.7] # Threshold for 'strong'
display(positive_corr.drop_duplicates().to_frame(name='Correlation'))

# Identify strongest negative relationships
print("\nStrongest Negative Relationships:")
negative_corr = corr_matrix.stack().sort_values(ascending=True)
negative_corr = negative_corr[negative_corr < -0.5] # Threshold for 'strong'
display(negative_corr.drop_duplicates().to_frame(name='Correlation'))

# Identify highly correlated input features (potential multicollinearity)
print("\nHighly Correlated Input Features (e.g., |correlation| > 0.7, excluding target):")
high_corr_inputs = corr_matrix.copy()
# Remove target variables from this check if they are in `correlation_vars`
if 'max_monthly_emi' in high_corr_inputs.columns:
    high_corr_inputs = high_corr_inputs.drop(columns=['max_monthly_emi'])
if 'emi_eligibility_encoded' in high_corr_inputs.columns:
    high_corr_inputs = high_corr_inputs.drop(columns=['emi_eligibility_encoded'])

if 'max_monthly_emi' in high_corr_inputs.index:
    high_corr_inputs = high_corr_inputs.drop(index=['max_monthly_emi'])
if 'emi_eligibility_encoded' in high_corr_inputs.index:
    high_corr_inputs = high_corr_inputs.drop(index=['emi_eligibility_encoded'])

np.fill_diagonal(high_corr_inputs.values, 0) # Set diagonal to 0 for self-correlation
high_corr_inputs = high_corr_inputs.stack()
high_corr_inputs = high_corr_inputs[abs(high_corr_inputs) > 0.7] # Threshold
display(high_corr_inputs.drop_duplicates().to_frame(name='Correlation'))
```
