### Impute Missing Values in 'emi_eligibility' Column

As requested, the `emi_eligibility` column, which is the target for classification, has one missing value. To ensure data integrity and avoid issues in later stages (like `train_test_split` with `stratify`), we will impute this missing categorical value with the mode (most frequent value) of the column.

```python
# Calculate the mode of the 'emi_eligibility' column
mode_emi_eligibility = df['emi_eligibility'].mode()[0]

# Impute NaN values in 'emi_eligibility' with the calculated mode
df['emi_eligibility'] = df['emi_eligibility'].fillna(mode_emi_eligibility)

print(f"Missing values in 'emi_eligibility' imputed with mode: '{mode_emi_eligibility}'")
print(f"New count of missing values in 'emi_eligibility': {df['emi_eligibility'].isnull().sum()}")
print(f"Updated unique values in 'emi_eligibility': {df['emi_eligibility'].unique()}")
```

### Define X and y for Classification and Regression Tasks

```python
# Define target variables
classification_target = 'emi_eligibility'
regression_target = 'max_monthly_emi'

# Identify features to exclude from X
excluded_features = [classification_target, regression_target]

# Create X and y for Classification
X_classification = df.drop(columns=excluded_features)
y_classification = df[classification_target]

# Create X and y for Regression
X_regression = df.drop(columns=excluded_features)
y_regression = df[regression_target]

print("X and y datasets created for classification and regression.")
```

### Phase 5.3 — Train / Validation / Test Split

```python
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42
TEST_SIZE = 0.30 # 30% for temp (validation + test)
VAL_TEST_SIZE_RATIO = 0.5 # 15% / 15% split from the 30% temp data

print("--- Classification Data Split ---")
# First split: 70% train, 30% temp (for validation and test)
X_class_train, X_class_temp, y_class_train, y_class_temp = train_test_split(
    X_classification_final, y_classification,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y_classification
)

# Second split: 15% validation, 15% test from the temp data
X_class_val, X_class_test, y_class_val, y_class_test = train_test_split(
    X_class_temp, y_class_temp,
    test_size=VAL_TEST_SIZE_RATIO,
    random_state=RANDOM_STATE,
    stratify=y_class_temp
)

print("Classification splits created.")

print("\n--- Regression Data Split ---")
# First split: 70% train, 30% temp (for validation and test)
X_reg_train, X_reg_temp, y_reg_train, y_reg_temp = train_test_split(
    X_regression_final, y_regression,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE
)

# Second split: 15% validation, 15% test from the temp data
X_reg_val, X_reg_test, y_reg_val, y_reg_test = train_test_split(
    X_reg_temp, y_reg_temp,
    test_size=VAL_TEST_SIZE_RATIO,
    random_state=RANDOM_STATE
)

print("Regression splits created.")

print("\n--- Displaying Shapes of Resulting Datasets ---")
print("\nClassification Datasets Shapes:")
print(f"X_class_train shape: {X_class_train.shape}")
print(f"y_class_train shape: {y_class_train.shape}")
print(f"X_class_val shape: {X_class_val.shape}")
print(f"y_class_val shape: {y_class_val.shape}")
print(f"X_class_test shape: {X_class_test.shape}")
print(f"y_class_test shape: {y_class_test.shape}")

print("\nRegression Datasets Shapes:")
print(f"X_reg_train shape: {X_reg_train.shape}")
print(f"y_reg_train shape: {y_reg_train.shape}")
print(f"X_reg_val shape: {X_reg_val.shape}")
print(f"y_reg_val shape: {y_reg_val.shape}")
print(f"X_reg_test shape: {X_reg_test.shape}")
print(f"y_reg_test shape: {y_reg_test.shape}")
```

### Phase 5.4 — Encoding — Part 2: Encoding Strategy Documentation

Based on the identified categorical features, here is the proposed encoding strategy for both classification and regression tasks. The same strategy will apply to both, and no actual encoding will be performed at this stage.

| Feature | Categories (from inspection) | Encoding Method |
|:---|:---:|:---:|

### Phase 5.4 — Encoding — Part 3: Implementing Encoding Strategy

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder

# Define binary/ordinal features and their mappings (as previously defined)
binary_ordinal_features = ['gender', 'marital_status', 'existing_loans']
binary_ordinal_mappings = {
    'gender': {'F': 0, 'M': 1},
    'marital_status': {'Single': 0, 'Married': 1},
    'existing_loans': {'No': 0, 'Yes': 1}
}

# Define one-hot features for classification (excluding 'emi_scenario' as per Phase 5.2 decision)
one_hot_features_class = ['education', 'employment_type', 'company_type', 'house_type', 'emi_scenario']
# Define one-hot features for regression (including 'emi_scenario')
one_hot_features_reg = ['education', 'employment_type', 'company_type', 'house_type', 'emi_scenario']

# --- Encoding for Classification Datasets ---

print("\n--- Encoding Classification Datasets ---")

# Make copies to work on, preserving original indices
X_class_train_encoded = X_class_train.copy()
X_class_val_encoded = X_class_val.copy()
X_class_test_encoded = X_class_test.copy()

# Apply binary/ordinal encoding using replace
for feature, mapping in binary_ordinal_mappings.items():
    if feature in X_class_train_encoded.columns:
        X_class_train_encoded[feature] = X_class_train_encoded[feature].replace(mapping)
        X_class_val_encoded[feature] = X_class_val_encoded[feature].replace(mapping)
        X_class_test_encoded[feature] = X_class_test_encoded[feature].replace(mapping)

# Initialize and fit OneHotEncoder on training data for classification
ohe_class = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
ohe_class.fit(X_class_train_encoded[one_hot_features_class])

# Get transformed feature names for OHE
ohe_class_feature_names = ohe_class.get_feature_names_out(one_hot_features_class)

# Transform datasets
X_class_train_ohe = ohe_class.transform(X_class_train_encoded[one_hot_features_class])
X_class_val_ohe = ohe_class.transform(X_class_val_encoded[one_hot_features_class])
X_class_test_ohe = ohe_class.transform(X_class_test_encoded[one_hot_features_class])

# Convert OHE arrays to DataFrames, preserving original indices
X_class_train_ohe_df = pd.DataFrame(X_class_train_ohe, columns=ohe_class_feature_names, index=X_class_train_encoded.index)
X_class_val_ohe_df = pd.DataFrame(X_class_val_ohe, columns=ohe_class_feature_names, index=X_class_val_encoded.index)
X_class_test_ohe_df = pd.DataFrame(X_class_test_ohe, columns=ohe_class_feature_names, index=X_class_test_encoded.index)

# Drop original OHE features and concatenate with the new OHE features
X_class_train_encoded = X_class_train_encoded.drop(columns=one_hot_features_class)
X_class_val_encoded = X_class_val_encoded.drop(columns=one_hot_features_class)
X_class_test_encoded = X_class_test_encoded.drop(columns=one_hot_features_class)

X_class_train_encoded = pd.concat([X_class_train_encoded, X_class_train_ohe_df], axis=1)
X_class_val_encoded = pd.concat([X_class_val_encoded, X_class_val_ohe_df], axis=1)
X_class_test_encoded = pd.concat([X_class_test_encoded, X_class_test_ohe_df], axis=1)

# Ensure column order is consistent across all splits
X_class_val_encoded = X_class_val_encoded[X_class_train_encoded.columns]
X_class_test_encoded = X_class_test_encoded[X_class_train_encoded.columns]


# --- Encoding for Regression Datasets ---

print("\n--- Encoding Regression Datasets ---")

# Make copies to work on, preserving original indices
X_reg_train_encoded = X_reg_train.copy()
X_reg_val_encoded = X_reg_val.copy()
X_reg_test_encoded = X_reg_test.copy()

# Apply binary/ordinal encoding using replace
for feature, mapping in binary_ordinal_mappings.items():
    if feature in X_reg_train_encoded.columns:
        X_reg_train_encoded[feature] = X_reg_train_encoded[feature].replace(mapping)
        X_reg_val_encoded[feature] = X_reg_val_encoded[feature].replace(mapping)
        X_reg_test_encoded[feature] = X_reg_test_encoded[feature].replace(mapping)

# Initialize and fit OneHotEncoder on training data for regression
ohe_reg = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
ohe_reg.fit(X_reg_train_encoded[one_hot_features_reg])

# Get transformed feature names for OHE
ohe_reg_feature_names = ohe_reg.get_feature_names_out(one_hot_features_reg)

# Transform datasets
X_reg_train_ohe = ohe_reg.transform(X_reg_train_encoded[one_hot_features_reg])
X_reg_val_ohe = ohe_reg.transform(X_reg_val_encoded[one_hot_features_reg])
X_reg_test_ohe = ohe_reg.transform(X_reg_test_encoded[one_hot_features_reg])

# Convert OHE arrays to DataFrames, preserving original indices
X_reg_train_ohe_df = pd.DataFrame(X_reg_train_ohe, columns=ohe_reg_feature_names, index=X_reg_train_encoded.index)
X_reg_val_ohe_df = pd.DataFrame(X_reg_val_ohe, columns=ohe_reg_feature_names, index=X_reg_val_encoded.index)
X_reg_test_ohe_df = pd.DataFrame(X_reg_test_ohe, columns=ohe_reg_feature_names, index=X_reg_test_encoded.index)

# Drop original OHE features and concatenate with the new OHE features
X_reg_train_encoded = X_reg_train_encoded.drop(columns=one_hot_features_reg)
X_reg_val_encoded = X_reg_val_encoded.drop(columns=one_hot_features_reg)
X_reg_test_encoded = X_reg_test_encoded.drop(columns=one_hot_features_reg)

X_reg_train_encoded = pd.concat([X_reg_train_encoded, X_reg_train_ohe_df], axis=1)
X_reg_val_encoded = pd.concat([X_reg_val_encoded, X_reg_val_ohe_df], axis=1)
X_reg_test_encoded = pd.concat([X_reg_test_encoded, X_reg_test_ohe_df], axis=1)

# Ensure column order is consistent across all splits
X_reg_val_encoded = X_reg_val_encoded[X_reg_train_encoded.columns]
X_reg_test_encoded = X_reg_test_encoded[X_reg_train_encoded.columns]


# --- Reporting ---
print("\n--- Encoding Implementation Report ---")

# Classification Report
print("\nClassification Datasets:")
print(f"  Original feature count (X_classification_final): {X_classification_final.shape[1]}")
print(f"  Encoded feature count (X_class_train_encoded): {X_class_train_encoded.shape[1]}")
print(f"  Encoded feature names (first 5 for brevity): {list(X_class_train_encoded.columns[:5])}...")
print(f"  Shape of X_class_train_encoded: {X_class_train_encoded.shape}")
print(f"  Shape of X_class_val_encoded: {X_class_val_encoded.shape}")
print(f"  Shape of X_class_test_encoded: {X_class_test_encoded.shape}")
print(f"  Confirmation of consistent columns (train vs val): {X_class_train_encoded.columns.equals(X_class_val_encoded.columns)}")
print(f"  Confirmation of consistent columns (train vs test): {X_class_train_encoded.columns.equals(X_class_test_encoded.columns)}")

# Regression Report
print("\nRegression Datasets:")
print(f"  Original feature count (X_regression_final): {X_regression_final.shape[1]}")
print(f"  Encoded feature count (X_reg_train_encoded): {X_reg_train_encoded.shape[1]}")
print(f"  Encoded feature names (first 5 for brevity): {list(X_reg_train_encoded.columns[:5])}...")
print(f"  Shape of X_reg_train_encoded: {X_reg_train_encoded.shape}")
print(f"  Shape of X_reg_val_encoded: {X_reg_val_encoded.shape}")
print(f"  Shape of X_reg_test_encoded: {X_reg_test_encoded.shape}")
print(f"  Confirmation of consistent columns (train vs val): {X_reg_train_encoded.columns.equals(X_reg_val_encoded.columns)}")
print(f"  Confirmation of consistent columns (train vs test): {X_reg_train_encoded.columns.equals(X_reg_test_encoded.columns)}")
```

### Phase 5.6 — Classification Imbalance — Part 1: Analyzing Class Distribution

```python
print("\n--- Analyzing Classification Class Distribution ---")

import numpy as np

def analyze_class_distribution(y_data, name):
    counts = y_data.value_counts()
    percentages = y_data.value_counts(normalize=True) * 100
    total_samples = len(y_data)

    # Identify minority and majority classes
    minority_class = counts.index[-1]
    majority_class = counts.index[0]
    imbalance_ratio = counts[majority_class] / counts[minority_class]

    print(f"\n{name} Class Distribution:")
    print(f"Total Samples: {total_samples}")
    for cls in counts.index:
        print(f"  Class '{cls}': Count = {counts[cls]}, Percentage = {percentages[cls]:.2f}%")
    print(f"Minority Class: '{minority_class}' (Count: {counts[minority_class]})")
    print(f"Majority Class: '{majority_class}' (Count: {counts[majority_class]})")
    print(f"Imbalance Ratio (Majority / Minority): {imbalance_ratio:.2f}")
    return percentages # Return percentages for later verification

# Analyze each split
print("\n--- Classification Training Data ---")
train_percentages = analyze_class_distribution(y_class_train, "Training Set")

print("\n--- Classification Validation Data ---")
val_percentages = analyze_class_distribution(y_class_val, "Validation Set")

print("\n--- Classification Test Data ---")
test_percentages = analyze_class_distribution(y_class_test, "Test Set")

print("\n--- Verification of Representative Proportions ---")
# Check if validation and test proportions are close to training proportions
proportion_check_val_train = all(np.isclose(val_percentages[cls], train_percentages[cls], atol=0.01) for cls in train_percentages.index)
proportion_check_test_train = all(np.isclose(test_percentages[cls], train_percentages[cls], atol=0.01) for cls in train_percentages.index)

if proportion_check_val_train and proportion_check_test_train:
    print("Class proportions in Validation and Test sets are representative of the Training distribution.")
else:
    print("WARNING: Class proportions in Validation/Test sets differ significantly from Training distribution.")

print("\n--- Conclusion on Class Imbalance ---")
if train_percentages.min() < 5 or (train_percentages.max() / train_percentages.min() > 3):
    print("The classification dataset exhibits significant class imbalance. The minority class accounts for a very small percentage of the data, and the imbalance ratio (majority/minority) is high. This level of imbalance is likely to negatively impact model performance, especially for the minority class, and therefore justifies the use of a sampling technique (e.g., SMOTE for oversampling, or techniques like class weighting) in a later stage.")
else:
    print("The classification dataset shows a relatively balanced class distribution, or the imbalance is not severe enough to definitively require sampling techniques at this stage.")
```

### Phase 5.6 — Classification Imbalance — Part 2: Defining the Imbalance Handling Strategy

Based on the analysis of the class distribution and considering the characteristics of the dataset (severe imbalance in the 'High_Risk' class, large dataset size, and the need to preserve validation/test distributions), **Class Weighting** is selected as the primary strategy for the initial modeling pipeline.

#### Rationale:

1.  **Direct Imbalance Compensation**: Class weighting adjusts the learning algorithm to penalize misclassifications of minority classes more heavily. This forces the model to pay more attention to the underrepresented classes without physically altering the data.
2.  **Data Integrity Preservation**: This method does not involve generating synthetic samples (like SMOTE) or discarding existing samples (like undersampling). This means the original data distribution and relationships within the training set are preserved, reducing the risk of introducing artificial patterns or losing valuable information.
3.  **Reduced Overfitting Risk**: Since no synthetic samples are created, the risk of overfitting to duplicate or artificially generated minority samples is avoided. This is particularly important with highly imbalanced datasets where synthetic oversampling can sometimes lead to models that perform well on augmented data but poorly on unseen real data.
4.  **Computational Efficiency**: Class weighting is generally less computationally intensive than sampling techniques, as it operates by modifying the loss function rather than manipulating the dataset directly. This makes it a practical choice for larger datasets.
5.  **Validation and Test Set Purity**: Crucially, class weighting is applied only during the training phase. The validation and test sets remain untouched and representative of the true data distribution, which is essential for unbiased model evaluation.
6.  **Compatibility with ML Algorithms**: Most modern machine learning algorithms (e.g., Logistic Regression, Support Vector Machines, Decision Trees, Gradient Boosting models) have built-in support for class weighting parameters.

While more advanced sampling techniques like SMOTE might be explored in later optimization phases if class weighting proves insufficient, starting with class weighting provides a robust and less intrusive first step to address the severe class imbalance.
